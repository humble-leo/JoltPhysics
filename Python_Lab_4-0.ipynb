````python
skaicius = 0

if skaicius > 0:
  print (f"Skaicius {skaicius} yra teigiamas")
else:
  print (f"Skaicius {skaicius} yra neigiamas")
````

Pabandykite viršuje esančiame kode įvesti, kad skaičius lygus 0. Jis įvertinamas kaip neigiamas, nes 0 nėra daugiau už 0.

`else` kodas vykdomas tik tuo atveju, jeigu netenkinama jokia aukščiau pateikta sąlyga.

#### A2 - Pirmas valymas: neįmanomos vertės

Laikas `0.0` reiškia, kad jutiklis nesuveikė. Tokią vertę reikia pašalinti **prieš** bet kokį skaičiavimą - ji ne „bloga“, ji apskritai nėra matavimas.

1. Pereikite per matavimus **su jų numeriais**: `for i in range(len(laikai)):`
2. Jei `laikai[i] <= 0`, atspausdinkite matavimo numerį (žmonės skaičiuoja nuo 1, Python - nuo 0) ir padidinkite skaitliuką `atmesta`.
3. Kitu atveju pridėkite vertę į naują sąrašą `tikri`.
4. Po ciklo atspausdinkite, kiek atmesta ir kiek liko.

Čia yra **du** skaitliukai - sąrašas ir skaičius. Abu paruošiami prieš ciklą.

*Turi pasirodyti:*
```
matavimas nr. 9: 0.0 s - jutiklis nesuveikė
atmesta: 1, liko: 15
```
Jei gavote `nr. 8` - pamiršote, kad Python skaičiuoja nuo nulio.

````python
atmesta=0
tikri = []
for i in range(len(laikai)):
  if laikai[i] <= 0:
    print(f"matavimas nr. {i+1}: {laikai[i]} s - jutiklis nesuveikė")
    atmesta += 1
  else:
    tikri.append(laikai[i])
print(f"atmesta: {atmesta}, liko: {len(tikri)}")
````

#### A3 - Vidurkis ir σ

Iš sąrašo `tikri` suskaičiuokite vidurkį, o **antru ciklu** - standartinį nuokrypį:

$\sigma = \sqrt{\dfrac{1}{N}\sum (t_i - \overline{t})^2}$

Vidurkį spausdinkite su 4 skaitmenimis po kablelio, σ - su 5.

*Turi pasirodyti:* `vidurkis = 0.4977 s`, `sigma = 0.01163 s`

Jei vidurkis išėjo `0.0330` - skaitliukas nunulinamas ciklo viduje. Patikrinkite kodo perstumimą.

Klausimas atsakymui komentare: beveik visi laikai skiriasi tik ketvirtame skaitmenyje po kablelio. Kodėl tada σ išėjo apie 0,01 s?

````python
from math import sqrt
vidurkis = sum(tikri) / len(tikri)
print(f"vidurkis = {vidurkis:.4f} s")

sigma = 0

sigma = sqrt((1/len(tikri))*sum((tikri[0]-vidurkis)**2))
````
